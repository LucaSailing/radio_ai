MODEL CNN

```python
df_cnn = df_filtered_CNN.copy()

# Transformer NaN en 0
# 1 = fracture visible
# 0 = pas de fracture visible
df_cnn["fracture"] = df_cnn["fracture_visible"].fillna(0).astype(int)

print("Nombre total :", len(df_cnn))
print(df_cnn["fracture"].value_counts())
```

```python
import os
import glob
import zipfile
import shutil

TARGET_DIR = "notebooks_modibo/cnn_images"

# Supprimer l'ancien dossier
if os.path.exists(TARGET_DIR):
    shutil.rmtree(TARGET_DIR)

os.makedirs(TARGET_DIR, exist_ok=True)

# Liste des images nécessaires
wanted = set(df_cnn["filestem"].astype(str) + ".png")

zip_files = glob.glob("/content/images_part*.zip")

found = set()

for zip_path in zip_files:

    print("Lecture :", zip_path)

    with zipfile.ZipFile(zip_path, "r") as z:

        for member in z.namelist():

            filename = os.path.basename(member)

            if filename in wanted and filename not in found:

                destination = os.path.join(TARGET_DIR, filename)

                with z.open(member) as source, open(destination, "wb") as target:
                    shutil.copyfileobj(source, target)

                found.add(filename)

print("\nImages demandées :", len(wanted))
print("Images retrouvées :", len(found))
print("Images manquantes :", len(wanted - found))
```

```python
df_cnn = df_filtered_CNN.copy()

df_cnn["fracture"] = (
    df_cnn["fracture_visible"]
    .fillna(0)
    .astype(int)
)

df_cnn["file_path"] = (
    "/content/cnn_images/"
    + df_cnn["filestem"].astype(str)
    + ".png"
)

print(len(df_cnn))
print(df_cnn["fracture"].value_counts())
print(df_cnn["file_path"].apply(os.path.exists).sum())
```

```python
from sklearn.model_selection import GroupShuffleSplit

# 70 % train / 30 % temporaire
split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(
    split1.split(
        df_cnn,
        y=df_cnn["fracture"],
        groups=df_cnn["patient_id"]
    )
)

train_df = df_cnn.iloc[train_idx].copy()
temp_df = df_cnn.iloc[temp_idx].copy()

# Séparer les 30 % restants en 15 % validation / 15 % test
split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(
    split2.split(
        temp_df,
        y=temp_df["fracture"],
        groups=temp_df["patient_id"]
    )
)

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()

print("TRAIN :", len(train_df))
print(train_df["fracture"].value_counts())

print("\nVALIDATION :", len(val_df))
print(val_df["fracture"].value_counts())

print("\nTEST :", len(test_df))
print(test_df["fracture"].value_counts())

print("\nTOTAL :", len(train_df) + len(val_df) + len(test_df))
```

```python
import tensorflow as tf

IMG_SIZE = (256, 256)
BATCH_SIZE = 16

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_png(image, channels=1)
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

def make_dataset(df, training=False):
    ds = tf.data.Dataset.from_tensor_slices(
        (df["file_path"].values, df["fracture"].values)
    )

    if training:
        ds = ds.shuffle(len(df))

    return (ds
            .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(tf.data.AUTOTUNE))

train_ds = make_dataset(train_df, True)
val_ds = make_dataset(val_df)
test_ds = make_dataset(test_df)
```

```python
from tensorflow.keras import Sequential, layers

model = Sequential([
    layers.Input(shape=(256, 256, 1)),

    layers.Conv2D(
        16,
        (3, 3),
        padding="same",
        activation="relu"
    ),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(
        32,
        (2, 2),
        padding="same",
        activation="relu"
    ),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Flatten(),

    layers.Dense(50, activation="relu"),

    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs= 10
)
```
